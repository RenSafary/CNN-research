```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("msambare/fer2013")

print("Path to dataset files:", path)
```

```python
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms

import os
from PIL import Image
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
device
```

```python
class Emotions(Dataset):
    def __init__(self, path_dir, transform=None):
        self.classes = {
            "angry": 0,
            "disgust": 1,
            "fear": 2,
            "happy": 3,
            "neutral": 4,
            "sad": 5,
            "surprise": 6
        }
        self.transform = transform
        self.samples = []

        for class_name, label in self.classes.items():
            folder = os.path.join(path_dir, class_name)

            for filename in sorted(os.listdir(folder)):
                path = os.path.join(folder, filename)
                self.samples.append((path, label))

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, i):
        img_path, label = self.samples[i]

        img = Image.open(img_path).convert("L")

        if self.transform:
            img = self.transform(img)

        return img, torch.tensor(label)
```

```python
trasform_train = transforms.Compose([
    transforms.RandomRotation(15),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.5],
        std=[0.5]
    ),
])

transform_test = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.5],
        std=[0.5]
    )
])
```

```python
train = "data/train"
test = "data/test"

emotions_train_dataset = Emotions(train, trasform_train)
emotions_test_dataset = Emotions(test, transform_test)
```

```python
train_loader = DataLoader(emotions_train_dataset, batch_size=64, shuffle=True)
test_loader = DataLoader(emotions_test_dataset, batch_size=64, shuffle=False)
```

```python
import numpy as np

labels_list = [label for _, label in emotions_train_dataset.samples]
class_counts = np.bincount(labels_list)

num_classes = 7
total_samples = len(labels_list)
```

```python
import matplotlib.pyplot as plt

emotions = ["angry", "disgust", "fear", "happy", "neutral", "sad", "surprise"]

plt.bar(emotions, class_counts)
plt.title("Dataset")
plt.xlabel("Classes")
plt.ylabel("Count")
```

```python
class CNN(nn.Module):
    def __init__(self):
        super().__init__()

        self.features = nn.Sequential(
            nn.Conv2d(1, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),

            nn.Conv2d(64, 128, 3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),

            nn.MaxPool2d(2),

            # 24x24
            nn.Conv2d(128, 256, 3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(),

            nn.Conv2d(256, 256, 3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(),

            nn.MaxPool2d(2)
        )

        self.classifier = nn.Sequential(
            nn.Flatten(),

            nn.Linear(256 * 12 * 12, 128),
            nn.ReLU(),
            nn.Dropout(0.3),

            nn.Linear(128, 7)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)

        return x
```

```python
model = CNN()
model = model.to(device)


class_weights = total_samples / (num_classes * class_counts)
weights_tensor = torch.tensor(class_weights, dtype=torch.float32).to(device)


loss_func = nn.CrossEntropyLoss(weight=weights_tensor)
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.0001,
    weight_decay=1e-4 # penalties for big weights (to avoid learning pictures)
)
```

```python
epochs = 100

x = []
y = []

for epoch in range(epochs):
    running_loss = 0.0

    for inputs, labels in train_loader:
        inputs = inputs.to(device)
        labels = labels.to(device)

        outputs = model(inputs)
        loss = loss_func(outputs, labels)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        running_loss += loss.item()

    epoch_loss = running_loss / len(train_loader)

    if (epoch + 1) % 10 == 0:
        print(f"Epoch: {epoch + 1}/{epochs} | Avg Loss: {epoch_loss:.4f}")

    x.append(epoch + 1)
    y.append(epoch_loss)
```

```python
plt.plot(x, y)
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.show()
```

```python
correct, total = 0, 0

model.eval()
with torch.no_grad():
    for inputs, labels in test_loader:
        inputs, labels = inputs.to(device), labels.to(device)
        prediction = model(inputs)
        _, predicted = torch.max(prediction, dim=1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

accuracy = correct / total
print(f"Test Accuracy: {accuracy*100:.2f}%")
```
